## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR", "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR", "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE", "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE", "Monitor 27", 25, 142.00],
]

print("Nombre producto de la 4ta fila:", inventario[3][1])

valor_stock = []
for fila in inventario:
    valor = round(fila[2] * fila[3], 2)
    valor_stock.append([fila[1], valor])

print("Valor stock:", valor_stock)

valor_inmovilizado = sum([f[1] for f in valor_stock])
print("Valor inmovilizado total:", round(valor_inmovilizado, 2))

inventario_sur = inventario[2:4]
valor_sur = round(sum([f[2] * f[3] for f in inventario_sur]), 2)
print("Valor inmovilizado ALM-SUR:", valor_sur)
```

Problema al usar posiciones fijas:
Si los datos de los almacenes no están ordenados o seguidos, sacar la información contando las posiciones fallará y sacará datos de otros lados. Lo más seguro en estos casos es usar una condición (un if) en vez de posiciones fijas.

## Ejercicio 3 — Manipulación destructiva y copias de listas

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00]

precios_rebajados = precios_originales
precios_rebajados[0] = 49.90
print("Originales:", precios_originales)
print("Rebajados:", precios_rebajados)
print("Explicación: Al usar '=', Python no copia los datos reales, sino que crea un enlace a la misma lista original en la memoria.")

precios_originales[0] = 59.90

copia_list = list(precios_originales)
copia_slice = precios_originales[:]
copia_list[0] = 99.90
print("Originales tras tocar copia:", precios_originales)

for i in range(len(copia_slice)):
    copia_slice[i] = round(copia_slice[i] * 0.9, 2)
print("Copia rebajada 10%:", copia_slice)

copia_slice.extend([10.00, 20.00])
copia_slice.insert(2, 100.00)

del copia_slice[4]
descartado = copia_slice.pop()
print("Descartado:", descartado)

print("ID original:", id(precios_originales))
print("ID copia_list:", id(copia_list))
print("ID copia_slice:", id(copia_slice))
```

## Ejercicio 5 — Funciones propias para el cálculo de importes

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """Devuelve el importe neto redondeado a dos decimales."""
    return round((unidades * precio_unitario) * (1 - descuento_pct / 100), 2)

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """Devuelve una tupla: (importe_bruto, ahorro, importe_neto)."""
    bruto = unidades * precio_unitario
    neto = calcular_importe(unidades, precio_unitario, descuento_pct)
    ahorro = bruto - neto
    return round(bruto, 2), round(ahorro, 2), neto

bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print(f"Bruto: {bruto}, Ahorro: {ahorro}, Neto: {neto}")

help(calcular_importe)

pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = [calcular_importe(*p) for p in pedidos]
print("Max:", max(importes))
print("Min:", min(importes))
print("Sorted:", sorted(importes))
print("Len:", len(importes))
```

## Ejercicio 6 — Normalización de códigos SKU con métodos

```python
skus_crudos = [
    " elc-0012-es ", "HOG-0045-ES", "dep-0003-pt ",
    "ofi-0021-es", " ELC-0012-ES", "hog-0099-fr ",
]

skus_limpios = [s.strip().upper() for s in skus_crudos]

print("Apariciones de ELC-0012-ES:", skus_limpios.count("ELC-0012-ES"))
print("Posición de OFI-0021-ES:", skus_limpios.index("OFI-0021-ES"))

partes = [s.split("-") for s in skus_limpios]
print("SKU en partes:", partes)

skus_es = [s.replace("-", ".") for s in skus_limpios if s.endswith("-ES")]
print("SKUs ES modificados:", skus_es)

cadena_unica = " | ".join(skus_limpios)
print("Cadena única:", cadena_unica)
```

## Ejercicio 10 — Matriz de ventas por trimestre y región

```python
import numpy as np

ventas_2d = np.array([
    [15200.50, 11800.25, 12950.00, 9870.40, 13400.10],
    [16840.75, 12310.60, 13480.90, 11020.35, 14100.80],
    [12535.72, 10663.06, 11221.59, 12673.08, 12334.38],
])

print("Shape:", ventas_2d.shape, "Ndim:", ventas_2d.ndim, "Size:", ventas_2d.size)
# Shape 3x5 indica trimestres y regiones. Ndim=2 es el número de dimensiones. Size=15 son las celdas totales.

print("Fila T2:", ventas_2d[1, :])
print("Columna Centro:", ventas_2d[:, 4])

print("Valor T3, Este:", ventas_2d[2, 2])

print("Submatriz T1-T2, primeras 3 regiones:\n", ventas_2d[:2, :3])

totales_trimestre = ventas_2d.sum(axis=1)
totales_region = ventas_2d.sum(axis=0)
print("Totales trimestre:", totales_trimestre)
print("Totales región:", totales_region)

fila_t4 = ventas_2d[2, :] * 1.06
ventas_ampliada = np.vstack([ventas_2d, fila_t4])
print("Shape ampliado:", ventas_ampliada.shape)

etiquetas_regiones = ["Norte", "Sur", "Este", "Oeste", "Centro"]
idx_max = np.argmax(totales_region)
print("Región con mayor facturación:", etiquetas_regiones[idx_max])
```
